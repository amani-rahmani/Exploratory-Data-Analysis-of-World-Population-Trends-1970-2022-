```python
years = [
    "1970 Population", "1980 Population", "1990 Population",
    "2000 Population", "2010 Population", "2015 Population",
    "2020 Population", "2022 Population"
]

top10 = df.sort_values("2022 Population", ascending=False).head(10)

plt.figure(figsize=(12, 6))
for _, row in top10.iterrows():
    plt.plot(years, row[years], label=row["Country"])

plt.title("Population Growth (1970–2022): Top 10 Countries")
plt.xlabel("Year")
plt.ylabel("Population")
plt.legend(bbox_to_anchor=(1.05, 1))
plt.tight_layout()
plt.show()
```

```python
continent_pop = (
    df.groupby("Continent")["2022 Population"]
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(8, 5))
continent_pop.plot(kind="bar")
plt.title("Total Population by Continent (2022)")
plt.xlabel("Continent")
plt.ylabel("Population")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(
    df["Density (per km²)"],
    df["Growth Rate"],
    alpha=0.6
)

plt.title("Population Density vs Growth Rate")
plt.xlabel("Density (people per km²)")
plt.ylabel("Growth Rate")
plt.show()
```

```python
top15_share = (
    df.sort_values("World Population Percentage", ascending=False)
    .head(15)
)

plt.figure(figsize=(10, 5))
plt.barh(top15_share["Country"], top15_share["World Population Percentage"])
plt.gca().invert_yaxis()
plt.title("Top 15 Countries by Share of World Population")
plt.xlabel("World Population Percentage")
plt.show()
```

```python

heatmap_data = (
    df.groupby("Continent")[years]
    .sum()
)

plt.figure(figsize=(20, 6))
sns.heatmap(
    heatmap_data,
    annot=True,
    fmt=".0f",
    cmap="YlOrRd"
)

plt.title("Population Growth by Continent (1970–2022)")
plt.show()
```
